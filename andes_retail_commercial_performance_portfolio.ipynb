# 📊 Commercial Performance Dashboard 2024–2025

**Andes Retail Group**  
**Tool:** Tableau Public  
**Focus:** Commercial performance, customer segments, product categories, geography, and seasonality

This portfolio project analyzes Andes Retail Group's commercial performance across 2024 and 2025 through two complementary interactive dashboards. The objective is to identify changes in revenue, margin, order volume, Average Order Value, product mix, geographic performance, and seasonality, and to translate those findings into clear business insights.

🔗 **Interactive dashboard:**  
https://public.tableau.com/app/profile/rodrigo.garza.garc.a/viz/Andes_Retail_Commercial_Performance_2024_2025/Dashboard-Overview


## 1. Business Objective

The analysis addresses two complementary questions:

1. **How did commercial performance evolve between 2024 and 2025?**
2. **Which geographic, seasonal, customer, and product patterns are associated with the observed changes?**

The project was designed for both executive review and deeper analytical exploration.


## 2. Dashboard Design

### Dashboard - Overview

The executive dashboard summarizes overall business performance and highlights the main commercial patterns.

**KPIs**
- **Revenue:** overall commercial performance.
- **Margin:** profitability based on `Revenue - Cost`.
- **Orders:** transaction volume.
- **Average Order Value:** average revenue generated per order.

**Visualizations**
- **Monthly Revenue:** compares monthly performance across 2024 and 2025.
- **Segment Performance:** compares revenue and margin across Economy, Standard, and Premium customer segments.
- **Category Performance:** compares revenue contribution across product categories.

**Filters and interactions**
- Year.
- Customer Segment.
- Monthly Revenue, Segment Performance, and Category Performance can be used as interactive filters that update the KPIs and other Overview charts.

### Dashboard - Detail

The detailed dashboard explores the geographic, seasonal, and product dimensions behind the executive results.

**Visualizations**
- **Regional Performance:** compares North, Central, and South.
- **Country Performance:** compares the countries included in the dataset.
- **Seasonal Performance:** identifies seasonal differences in revenue.
- **Category by Season:** shows how category performance varies by season.
- **Order Details:** provides transaction-level detail for the selected filters.

**Filters and interactions**
- Order Date.
- Customer Segment.
- Product Category.
- Region.
- Region, Country, Season, and Category by Season can be used as interactive filters.


## 3. Key Performance Results

| Metric | 2024 | 2025 | Change |
|---|---:|---:|---:|
| Revenue | 2,863,384 | 2,668,610 | **-6.8%** |
| Margin | 1,005,811.62 | 935,937.71 | **-6.9%** |
| Orders | 2,550 | 2,450 | **-3.9%** |
| Average Order Value | 1,122.90 | 1,089.23 | Decreased |

Across both years, Andes Retail Group generated **5,531,994 in cumulative revenue**, **1,941,749.33 in total margin**, and **5,000 orders**.

The 2025 decline was therefore not concentrated in a single metric: revenue, margin, orders, and Average Order Value all moved downward.


## 4. Key Findings

### Product mix
- **Home moved from the lowest-revenue category in 2024 to the highest-revenue category in 2025.**
- Sports lost its leading position in 2025.
- The shift in category leadership is one of the most important changes requiring further investigation.

### Customer segments
- **Economy** consistently generated the lowest revenue and margin.
- **Standard and Premium** accounted for most of the commercial performance.

### Geography
- **North:** 1,890,166 in revenue.
- **Central:** 1,857,015.
- **South:** 1,784,813.
- By country, **Peru (2,158,578)** generated the highest revenue, followed by **Chile (2,027,669)** and **Colombia (1,345,747)**.

### Seasonality
- **Summer:** 2,241,719 in revenue.
- **Winter:** 653,513.
- Summer generated **more than three times** the revenue of Winter.
- All product categories reached their highest revenue during Summer.
- Monthly performance was weaker between **June and August**, followed by a rebound toward year-end.


## 5. Executive Narrative — SCQA

### Situation
During 2024 and 2025, Andes Retail Group generated **5,531,994 in cumulative revenue**, **1,941,749.33 in total margin**, and **5,000 orders**. Monthly performance followed a broadly similar pattern in both years, with stronger activity at the beginning and end of the year and lower revenue levels between June and August.

### Complication
Commercial performance declined in 2025. Revenue decreased by approximately **6.8%**, margin by **6.9%**, and orders by **3.9%**, while Average Order Value fell from **1,122.90** to **1,089.23**. At the same time, the product mix changed substantially, with **Home moving from the lowest-revenue category in 2024 to the highest in 2025**.

### Question
Which geographic, seasonal, customer, and product patterns are associated with the differences observed in commercial performance, and which changes require further analysis?

### Answer
The decline was broad-based rather than isolated to a single KPI. The detailed analysis also shows strong seasonality, moderate geographic differences, and a meaningful change in category leadership. In particular, Home's rise should be investigated further to determine whether it is associated with order volume, Average Order Value, seasonality, region, or other factors. These relationships should be interpreted as associations rather than causal effects.


## 6. Recommended Next Steps

- Investigate the drivers behind **Home's change in category leadership**.
- Review the factors associated with the **June–August slowdown**.
- Analyze opportunities to strengthen performance during **lower-activity periods**, especially Winter.
- Compare category performance by region, season, order volume, and Average Order Value before drawing causal conclusions.
- Continue monitoring the Economy segment to understand whether its lower revenue and margin reflect customer mix, transaction volume, or purchasing behavior.


## 7. Stakeholder Update

📊 **Commercial performance update**

Andes Retail Group closed 2024–2025 with **5.53M in revenue, 1.94M in margin, and 5,000 orders**.

In 2025, revenue declined by approximately **6.8%**, margin by **6.9%**, and orders by **3.9%** versus 2024. Activity was weakest between **June and August**, while year-end showed a rebound.

One of the most relevant changes was the product mix: **Home moved from the lowest-revenue category in 2024 to the highest in 2025**. In addition, the **Economy** segment consistently generated the lowest revenue and margin.

The detailed view also shows a strong seasonal effect: **Summer generated more than three times the revenue of Winter**, with all product categories reaching their highest revenue during Summer.

**Recommended next steps:** investigate the drivers behind Home's growth, review the factors associated with the mid-year slowdown, and identify opportunities to strengthen performance during the lowest-activity periods.
